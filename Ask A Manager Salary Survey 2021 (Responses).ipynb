```python
import re
import unicodedata
import pycountry
from rapidfuzz import process, fuzz
```

```python
col = "What country do you work in?"

def norm(x):
    x = unicodedata.normalize("NFKD", str(x)).encode("ascii", "ignore").decode()
    x = x.lower().strip()
    x = x.replace(".", "")
    x = re.sub(r"[^a-z\s]", " ", x)
    return re.sub(r"\s+", " ", x).strip()

uniques = df[col].dropna().unique()
normalization_map = {u: norm(u) for u in uniques}
```

```python
std_names = {
    "United States": "United States",
    **{c.name: c.name for c in pycountry.countries}
}

common = {c.common_name: c.name for c in pycountry.countries if hasattr(
    c, "common_name")}

std_names.update(common)

all_countries = list(std_names.keys())
```

```python
display_names = {
    "Viet Nam": "Vietnam",
    "Taiwan, Province of China": "Taiwan",
    "Iran, Islamic Republic of": "Iran",
    "Korea, Republic of": "South Korea",
    "Russian Federation": "Russia",
    "Türkiye": "Turkey",
    "Czechia": "Czech Republic",
}

def match_country(raw):
    key = normalization_map[raw]
    if not key:
        return None, "empty"
    if key in aliases:
        name, how = aliases[key], "alias"
    else:
        try:
            name, how = pycountry.countries.lookup(key).name, "pycountry"
        except LookupError:
            r = process.extractOne(key.title(), all_countries, scorer=fuzz.WRatio, score_cutoff=90)
            if not r:
                return None, "unmatched"
            name, how = std_names[r[0]], f"fuzzy({r[1]:.0f})"
    return display_names.get(name, name), how

results = {u: match_country(u) for u in uniques}
mapping = {u: r[0] for u, r in results.items()}
method  = {u: r[1] for u, r in results.items()}

df["Country"] = df[col].map(mapping)
```

```python
rules = [
    ("Education & Library",  r"instructional design|teacher|professor|instructor|librarian|archivist|lecturer|academic|school|faculty|tutor|curator"),
    ("Healthcare",           r"nurse|\brn\b|\bnp\b|doctor|physician|clinical|medical|therapist|pharmac|dental|health"),
    ("Social & Public Service", r"social worker|case manager|caseworker|counselor|program officer|policy"),
    ("Data & Analytics",     r"data|analyst|analytics|\bbi\b|statistic"),
    ("Software & IT",        r"software|developer|programmer|devops|\bit\b|systems|network|security|product owner|product manager|\bux\b|\bqa\b"),
    ("Engineering",          r"engineer"),
    ("Research & Science",   r"research|scientist|\blab\b|biolog|chemist"),
    ("Writing & Media",      r"editor|writer|copywriter|reporter|journalist|content|producer|communications"),
    ("Design",               r"design|illustrator|creative"),
    ("Marketing & Sales",    r"market|sales|account exec|account manager|brand|business development|\bpr\b"),
    ("Finance & Legal",      r"controller|accountant|accounting|finance|financial|auditor|attorney|lawyer|legal|counsel|paralegal|banker|tax"),
    ("Customer Service",     r"customer service|customer success|support specialist|client service"),
    ("HR",                   r"\bhr\b|human resources|recruit|talent"),
    ("Admin & Support",      r"admin|assistant|coordinator|receptionist|clerk|secretary"),
    ("Consulting",           r"consultant|advisor"),
    ("Operations & Project", r"operations|project|program manager|logistics|supply chain|procurement"),
    ("General Management",   r"manager|director|\bvp\b|president|chief|\bc[a-z]o\b|supervisor|head of|partner"),
]
```

```python
def update_rule(name, pattern):
    for i, (n, p) in enumerate(rules):
        if n == name:
            rules[i] = (n, pattern)
            return
    raise ValueError(f"Rule {name} does not exist")

update_rule("Education & Library",
    r"instructional design|teacher|professor|instructor|librarian|library|archivist|lecturer|academic|school|faculty|tutor|curator|student")

update_rule("Data & Analytics",
    r"data|analyst|analytics|\bbi\b|business intelligence|statistic")


update_rule("Software & IT",
    r"software|developer|programmer|devops|\bit\b|systems|network|security|product owner|product manager|\bqa\b")

update_rule("Design",
    r"design|illustrator|creative|\bux\b|art director")

update_rule("Marketing & Sales",
    r"market|sales|account exec|account manager|brand|business development|\bpr\b|social media")

update_rule("Finance & Legal",
    r"controller|accountant|accounting|finance|financial|auditor|attorney|lawyer|legal|counsel|paralegal|banker|tax|payroll")

update_rule("HR",
    r"\bhr\b|human resources|recruit|talent|people operations|people partner")

update_rule("Admin & Support",
    r"office manager|(?<!director of )admin|assistant(?! (?:\w+ )?(?:director|manager))|coordinator|receptionist|clerk|secretary")
```

```python
def job_family(title):
    if pd.isna(title):
        return np.nan
    t = title.lower()
    for name, pattern in rules:
        if re.search(pattern, t):
            return name
    return "Other"

df["Job title"] = df["Job title"].str.strip()
uniq = df["Job title"].dropna().unique()
fam_map = {u: job_family(u) for u in uniq}
df["job_family"] = df["Job title"].map(fam_map)

print(df["job_family"].value_counts(dropna=False))
```
